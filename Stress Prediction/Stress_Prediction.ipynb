```python
combined_train = pd.concat([time_domain_features_train, frequency_domain_features_train, heart_rate_non_linear_features_train], axis=1)
combined_train.head()
```

```python
combined_test = pd.concat([time_domain_features_test, frequency_domain_features_test, heart_rate_non_linear_features_test], axis=1)
combined_test.head()
```

```python
label_train = LabelEncoder()
combined_train['condition'] = label_train.fit_transform(combined_train['condition'])
```

```python
train_columns = set(combined_train.columns)
test_columns = set(combined_test.columns)

#columns that are in the training set but not in the test set
extra_columns_in_train = train_columns - test_columns


print("Columns in the training set but not in the test set:", extra_columns_in_train)
```

```python
combined_train = combined_train.drop(columns=['HR'])
```

```python
df_train = combined_train

#the target label
label = ['condition']

#the features to be used for model training
features = ['MEAN_RR','RMSSD','pNN25','pNN50','LF','HF','LF_HF']
```

```python
df_train[features].isnull().sum()
```

```python
scaler = StandardScaler()
scaler.fit(df_train[features])
df_train[features] = scaler.transform(df_train[features])
```

```python
X = df_train[features].values
Y = df_train[label].values
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=40)
```

```python
print(X_train.shape,Y_train.shape,X_test.shape,Y_test.shape)
```

```python
from sklearn.preprocessing import LabelEncoder
import tensorflow.keras as k

Y_train = k.utils.to_categorical(Y_train)
Y_test = k.utils.to_categorical(Y_test)

print("Shape y_train after one-hot encoding:", Y_train.shape)
print("Shape y_test after one-hot encoding:", Y_test.shape)
```

```python
print(X_train.shape,Y_train.shape,X_test.shape,Y_test.shape)
```

```python
df_train.head()
```

```python
model = Sequential()
```

```python
model.add(Dense(50, activation='relu', input_shape=(X_train.shape[1],)))

model.add(Dense(20, activation='relu'))

model.add(Dense(10, activation='relu'))

model.add(Dense(3, activation='softmax'))
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

history = model.fit(X_train, Y_train, validation_data=(X_test, Y_test), epochs=50, batch_size=1024, verbose=1)
```

```python
label_test = LabelEncoder()
combined_test['condition'] = label_test.fit_transform(combined_test['condition'])
```

```python
df_test = combined_test
# the target label
label = ['condition']

# the features to be used for model training
features = ['MEAN_RR','RMSSD','pNN25','pNN50','LF','HF','LF_HF']
```

MEAN_RR: Average time between consecutive heartbeats, reflecting heart rate variability.
RMSSD: Square root of the mean of successive RR interval differences, indicating parasympathetic activity.
pNN25: Percentage of RR intervals differing by more than 25 ms, showing short-term heart rate variability.
pNN50: Percentage of RR intervals differing by more than 50 ms, reflecting parasympathetic tone.
LF: Power of the low-frequency band, representing both sympathetic and parasympathetic activity.
HF: Power of the high-frequency band, primarily reflecting parasympathetic activity.
LF_HF: Ratio of low to high-frequency power, indicating the balance between sympathetic and parasympathetic activity.



```python
scaler = StandardScaler()
scaler.fit(df_test[features])
df_test[features] = scaler.transform(df_test[features])
```

```python
X = df_test[features].values
Y = df_test[label].values
```

```python
Y = k.utils.to_categorical(Y)
```

```python
loss,acc = model.evaluate(X,Y,verbose=1)


print("Test Loss:", loss)
print("Test Accuracy:", acc)
```
